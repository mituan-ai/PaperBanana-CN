## 2. Configure the two model connections

The example uses one Google API key for both roles, but the two saved connections remain independent. Replace either provider, URL, key, or model when your VLM and image service use different endpoints.

````python
import os
from getpass import getpass

os.environ["PB_GOOGLE_API_KEY"] = getpass("Google API key: ")
````

````python
!paperbanana-cn connections add \
  --role vlm --name "Colab VLM" --provider gemini \
  --model gemini-2.5-flash --api-key-env PB_GOOGLE_API_KEY

!paperbanana-cn connections add \
  --role image --name "Colab Image" --provider google_imagen \
  --model gemini-3-pro-image-preview --size-mode native_tier \
  --api-key-env PB_GOOGLE_API_KEY
````

## 3. Describe the method

````python
methodology = """\
Our framework has three stages. First, a retriever selects evidence from a
domain knowledge base. Second, a planner builds a typed reasoning graph whose
nodes represent observations, hypotheses, and decisions. Third, a verifier
checks every decision against the retrieved evidence and sends unsupported
claims back to the planner for revision. The verified graph is decoded into
the final prediction together with an auditable evidence trail.
"""

with open("method.txt", "w", encoding="utf-8") as handle:
    handle.write(methodology)
````

## 4. Generate a 16:9, 2K methodology figure

This performs paid model calls. Change the budget or resolution before running when required by your provider.

````python
!paperbanana-cn generate \
  --input method.txt \
  --caption "Evidence-grounded planning and verification workflow" \
  --aspect-ratio 16:9 \
  --resolution 2K \
  --iterations 1 \
  --budget 0.50 \
  --output-dir outputs
````

````python
from pathlib import Path
from IPython.display import Image, display

result = max(Path("outputs").glob("run_*/final_output.*"), key=lambda path: path.stat().st_mtime)
display(Image(filename=str(result)))
````
